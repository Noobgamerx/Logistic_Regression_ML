Imporing Libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

Importing the data set

```python
data =  pd.read_csv("/content/drive/My Drive/ML Lab/log_reg.csv")
df=pd.read_csv("/content/drive/My Drive/ML Lab/log_reg.csv")
print(df)
```

Defining X and Y and spliting data

```python
X = df[['Gender', 'Age', 'EstimatedSalary']]
Y = df[['Purchased']]

X_train, X_test, y_train, y_test = train_test_split( X, Y, test_size=0.25, random_state=42, stratify=Y )
```

Preprocessing the data

```python
#labeled encoding

gender_encoder = LabelEncoder()

X_train_encoded = X_train.copy() # Start with a copy of X_train
X_test_encoded = X_test.copy()   # Start with a copy of X_test

# Now encode the 'Gender' column within these new copies
X_train_encoded['Gender'] = gender_encoder.fit_transform(X_train_encoded['Gender'])
X_test_encoded['Gender'] = gender_encoder.transform(X_test_encoded['Gender']) # Use transform for test set

print("Gender Mapping :")
for idx, label in enumerate(gender_encoder.classes_):
  print(f"  {label} -> {idx}")
print("-" * 50)

# Scaling the numerical features
scaler = StandardScaler()

numerical_cols = ['Age', 'EstimatedSalary']

# Apply scaling to the numerical columns of the already-encoded dataframes
# The 'Gender' column is already 0/1 and typically not scaled by StandardScaler in this context.
X_train_Scaled = X_train_encoded.copy()
X_train_Scaled[numerical_cols] = scaler.fit_transform(X_train_encoded[numerical_cols])

X_test_Scaled = X_test_encoded.copy()
X_test_Scaled[numerical_cols] = scaler.transform(X_test_encoded[numerical_cols])

print("X_train_Scaled head:")
print(X_train_Scaled.head())
print("\nX_test_Scaled head:")
print(X_test_Scaled.head())
```

Model Training

```python
model = LogisticRegression(random_state = 42)
model.fit(X_train_Scaled, y_train.values.ravel())

y_pred = model.predict(X_test_Scaled)

```

Printing the output

```python
print(f"Model Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['Not Purchased (0)', 'Purchased (1)']))
```
